## Slovníček pojmů

**Definice a volání**

* **funkce** - pojmenovaný kus kódu, který se dá opakovaně použít
* **definice** - příkaz `def`, který funkci vytvoří a uloží pod jméno;
  nic tím nespustí
* **volání** - použití funkce se závorkami, `odpor_vrstvy(0.30, 0.18)`;
  teprve teď se tělo provede
* **tělo funkce** - odsazený blok pod `def`

**Vstupy a výstupy**

* **parametr** - jméno v definici funkce
* **argument** - konkrétní hodnota předaná při volání
* **pozicní argument** - přiřazuje se podle pořadí
* **pojmenovaný argument** - přiřazuje se podle jména, `R_se=0.10`
* **výchozí hodnota** - hodnota parametru, když argument nezadáme
* **signatura** - první řádek definice; říká, jak se funkce volá
* **anotace typu** - `d: float`, `-> float`; popis pro čtenáře a editor,
  Python ho nevynucuje
* **docstring** - text v trojitých uvozovkách na začátku těla; zobrazí ho `help()`
* **`return`** - předá hodnotu volajícímu a funkci ukončí
* **`None`** - vrací ho funkce, která nemá `return`

**Jmenný prostor a vedlejší efekty**

* **lokální jméno** - jméno vzniklé v těle funkce; po skončení zmizí
* **globální jméno** - jméno vně funkce; funkce ho může číst, přiřazením
  by vytvořila nové lokální
* **čistá funkce** - výsledek závisí jen na argumentech a nic mimo sebe nemění
* **vedlejší efekt** - cokoli dalšího: výpis, změna seznamu z argumentu, zápis
  do souboru
* **měnitelný argument** - seznam nebo slovník, který funkce může změnit
  volajícímu pod rukama
* **měnitelná výchozí hodnota** - past `def f(x=[])`; výchozí hodnota se vytvoří
  jednou při definici, proto se použije `None`

**Funkce jako hodnota**

* **funkce je hodnota** - dá se uložit do proměnné a předat jako argument
* **`lambda`** - funkce bez jména na jednom řádku, `lambda p: p["objem"]`
* **`key`** - parametr funkcí `sorted()`, `max()` a `min()`, kterému se předá
  funkce
* **`map()`** - aplikuje funkci na každý prvek kolekce; výsledek je líný objekt,
  proto `list(map(...))`
* **metoda** - funkce uložená u typu; objekt před tečkou je její první argument
  (`self`), takže `"abc".upper()` je `str.upper("abc")`
