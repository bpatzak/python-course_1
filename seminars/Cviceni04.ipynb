## Slovníček pojmů

**Větvení**

* **podmínka** - výraz, jehož pravdivost rozhoduje, který kód se provede
* **blok** - řádky odsazené pod `if`, `elif`, `else`, `case`; provedou se
  společně
* **odsazení** - mezery na začátku řádku (standardně 4); v Pythonu je součástí
  syntaxe
* **`if`** - provede blok, jen když podmínka platí
* **`else`** - blok pro případ, že žádná předchozí podmínka neplatila
* **`elif`** - další podmínka v řadě; provede se jen první splněná větev
* **vnořená podmínka** - `if` uvnitř bloku jiného `if`
* **podmíněný výraz** - `a if podminka else b`, volba hodnoty na jednom řádku

**Podmínky**

* **`and`, `or`, `not`** - logické operátory; přednost má `not`, pak `and`,
  pak `or`
* **řetězené porovnání** - `0 <= x <= L`, zápis intervalu
* **zkrácené vyhodnocení** - `and` a `or` přestanou počítat, jakmile je
  výsledek jasný
* **pravdivostní hodnota (truthiness)** - každá hodnota je pravdivá nebo
  nepravdivá; nepravdivé jsou `False`, `None`, `0`, `0.0` a prázdné kolekce
* **`is None`** - test chybějící hodnoty; na rozdíl od `not x` neplete `None`
  s nulou
* **`in` / `not in`** - je hodnota v kolekci?

**`match` / `case`**

* **`match`** - porovná jednu hodnotu s řadou vzorů
* **vzor (pattern)** - to, co stojí za `case`: literál, tvar tuple nebo slovníku
* **`|`** - více možností v jednom vzoru, `case "XC2" | "XC3":`
* **`case _:`** - zachytí cokoli, obdoba `else`
* **zachycení (capture)** - holé jméno ve vzoru hodnotu uloží, neporovná ji
* **guard** - podmínka za vzorem, `case (x, y) if x > 0:`

**Chyby**

* **`SyntaxError`** - typicky `=` místo `==` nebo chybějící dvojtečka
* **`IndentationError`** - chybí nebo nesedí odsazení bloku
