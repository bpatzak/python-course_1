## Slovníček pojmů

**Cykly**

* **cyklus** - opakované provedení bloku kódu
* **průchod (iterace)** - jedno provedení těla cyklu
* **`for`** - projde postupně všechny prvky kolekce
* **řídicí proměnná** - proměnná za `for`, v každém průchodu drží aktuální
  prvek; po skončení cyklu drží poslední hodnotu
* **`while`** - opakuje blok, dokud platí podmínka
* **nekonečný cyklus** - podmínka `while` se nikdy nestane nepravdivou;
  přerušení v Jupyteru tlačítkem Interrupt, v terminálu `Ctrl+C`
* **vnořený cyklus** - cyklus uvnitř jiného cyklu

**Pomocné funkce**

* **`range(start, stop, krok)`** - řada celých čísel; `stop` do ní nepatří
* **`enumerate(kolekce, start=1)`** - dvojice (pořadí, prvek)
* **`zip(a, b)`** - souběžný průchod více kolekcemi, skončí s nejkratší
* **`.items()`** - dvojice klíč - hodnota při průchodu slovníkem
* **`any()` / `all()`** - platí podmínka aspoň pro jeden prvek / pro všechny?

**Řízení průchodu**

* **`break`** - okamžitě ukončí cyklus
* **`continue`** - přeskočí zbytek těla a pokračuje dalším průchodem
* **`else` u cyklu** - provede se, jen když cyklus doběhl bez `break`
* **`pass`** - prázdný příkaz, když blok nemá nic dělat

**Vzory**

* **akumulátor** - proměnná připravená před cyklem a aktualizovaná v každém
  průchodu (součet, počet, nový seznam, slovník součtů)
* **`+=`** - zkratka za `x = x + ...`
* **list comprehension** - `[vyraz for x in kolekce if podminka]`, nový seznam
  jedním výrazem
* **dict / set comprehension** - totéž pro slovník `{k: v for ...}` a množinu
  `{x for ...}`
* **generátorový výraz** - `(vyraz for x in kolekce)`, typicky uvnitř `any()`,
  `all()`, `sum()`
* **off-by-one** - chyba o jedna, typicky špatná horní mez `range()`
