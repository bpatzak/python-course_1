## Slovníček pojmů

**Tabulka**

* **`DataFrame`** - tabulka; sloupce mají názvy, řádky index
* **`Series`** - jeden sloupec tabulky; pole NumPy s indexem
* **index** - popisky řádků, obvykle 0, 1, 2, ...
* **CSV** - textový soubor s tabulkou, hodnoty oddělené čárkou
* **`pd.read_csv("soubor.csv", parse_dates=[...])`** - načte CSV, vybrané
  sloupce jako čas
* **`head()`, `info()`, `describe()`** - první řádky, typy sloupců, statistika

**Výběr**

* **`df["a"]`, `df[["a", "b"]]`** - jeden sloupec, víc sloupců
* **`df.loc[radky, sloupce]`** - výběr podle popisků
* **`df.iloc[radky, sloupce]`** - výběr podle pořadí
* **`df[maska]`** - řádky, kde maska platí
* **`~maska`** - opak masky
* **`df.loc[maska, "a"] = hodnota`** - změna hodnot jen ve vybraných řádcích
* **`unique()`, `value_counts()`** - různé hodnoty, četnost hodnot
* **`idxmax()`** - popisek řádku s největší hodnotou

**Chybějící hodnoty**

* **`NaN`** - chybějící hodnota; `sum()` a `mean()` ji potichu přeskočí
* **`isna()`** - maska chybějících hodnot
* **`fillna()`, `dropna()`** - doplnění, vyhození chybějících hodnot

**Skupiny**

* **`groupby("a")["b"].mean()`** - průměr sloupce `b` pro každou hodnotu `a`
* **`agg(["min", "max", "mean"])`** - víc agregací najednou
* **`sort_values("a")`** - seřazení, vrací novou tabulku

**Čas**

* **`dt.hour`, `dt.date`, `dt.day_of_week`** - hodina, datum, den v týdnu
  (0 = pondělí)
* **`Timedelta`** - doba trvání, rozdíl dvou časů
* **`dt.total_seconds()`** - doba trvání v sekundách
