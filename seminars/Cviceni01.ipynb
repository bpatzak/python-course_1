# Cvičení 01 - Instalace a první spuštění Pythonu

Cíle cvičení:

* nainstalovat Python a editor
* vytvořit virtuální prostředí
* spustit skript `.py`
* spustit notebook `.ipynb`
* vyzkoušet Google Colab (notebook v cloudu)

Na konci cvičení najdete slovníček pojmů, které se v textu objevují.


## 0. PowerShell (jen Windows)

PowerShell je terminál, tedy okno, do kterého se píšou příkazy pro operační
systém. Budeme v něm instalovat a spouštět Python.

* otevřete PowerShell (klávesa Windows, napsat `powershell`, Enter)
* povolte spouštění skriptů `.ps1` - Windows je ve výchozím nastavení blokují
  a my je budeme potřebovat, např. pro ruční aktivaci virtuálního prostředí:

```
Set-ExecutionPolicy -Scope CurrentUser -ExecutionPolicy RemoteSigned
```

* nastavení potvrďte klávesou `A` (ano všem) nebo `Y`
* na macOS a Linuxu tento krok přeskočte, tam se používá terminál s Bash/Zsh


## 1. Instalace uv

* `uv` je jeden nástroj pro správu Pythonu, balíčků i virtuálních prostředí
* je rychlý a nemusíte nic hledat a instalovat ručně
* dokumentace: <https://docs.astral.sh/uv/>

Instalace na **Windows** (PowerShell):

```
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

Instalace na **macOS / Linux** (terminál):

```
curl -LsSf https://astral.sh/uv/install.sh | sh
```

Po instalaci:

* zavřete a znovu otevřete terminál, aby se načetla nová cesta k `uv`
* ověřte, že `uv` funguje:

```
uv --version
```

Příkaz vypíše číslo verze, např. `uv 0.9.7`.

## 2. Instalace Pythonu pomocí uv

* samotný Python si `uv` stáhne sám, nic nemusíte hledat na webu
* nainstalujte verzi 3.13:

```
uv python install 3.13
```

* seznam dostupných a nainstalovaných verzí:

```
uv python list
```

Příkaz spusťte i v případě, že už nějaký Python v počítači máte (např. z
python.org nebo z Anacondy). `uv` si spravuje vlastní instalace, takže se
nemusíme starat o konflikty verzí a všichni budeme mít stejné prostředí.


## 3. Instalace VS Code

VS Code je editor, ve kterém budeme psát skripty i otevírat notebooky.

**Windows / macOS:**

* stáhněte instalátor z <https://code.visualstudio.com/>
* spusťte ho a postupujte podle průvodce (další, další, hotovo)

**Linux (Ubuntu/Debian):**

```
sudo snap install code --classic
```

Po instalaci ještě přidejte dvě rozšíření (záložka *Extensions* v levém
panelu, ikona čtverečků):

* **Python**
* **Jupyter**

Usnadňují spouštění skriptů a práci s notebooky.


## 4. Adresář pro cvičení a virtuální prostředí

* virtuální prostředí je samostatná složka s vlastní kopií Pythonu a balíčky,
  oddělená od zbytku počítače - typicky jedno prostředí na jeden projekt
* díky tomu si projekty navzájem nepřepisují verze balíčků

Postup (v terminálu):

```
mkdir cviceni01
cd cviceni01
uv venv
uv pip install ipykernel
code .
```

Co jednotlivé příkazy udělaly:

* `mkdir cviceni01` a `cd cviceni01` - vytvoří adresář pro dnešní cvičení
  a přepne se do něj
* `uv venv` - vytvoří virtuální prostředí, vznikla nová složka `.venv`
* `uv pip install ipykernel` - do prostředí přidá balíček `ipykernel`, který
  VS Code potřebuje ke spouštění notebooků
* `code .` - otevře aktuální adresář v editoru VS Code (tečka znamená
  "tento adresář")

Prostředí nemusíte nijak aktivovat, příkaz `uv run` ho použije sám.


## 5. Spuštění skriptu `.py`

* skript je textový soubor s kódem Pythonu, který se vykoná celý najednou
  odshora dolů
* ve VS Code vytvořte v adresáři `cviceni01` soubor `zdravim.py` s obsahem:

```python
jmeno = input("Jak se jmenujete? ")
print("Dobrý den,", jmeno)
```

* skript spusťte v terminálu (terminál otevřete i přímo ve VS Code:
  *Terminal -> New Terminal*; pokud jste ho zavřeli, vraťte se nejdřív do
  adresáře příkazem `cd cviceni01`):

```
uv run zdravim.py
```

* skript se zastaví a čeká, dokud nenapíšete jméno a nestisknete Enter


## 6. Spuštění notebooku `.ipynb`

* notebook je soubor, ve kterém se střídají **textové** a **kódové** buňky
* buňku spustíte klávesami `Shift` + `Enter`
* buňky se spouštějí v tom pořadí, ve kterém je spustíte vy (ne nutně odshora
  dolů) - na to pozor, je to častý zdroj chyb

**Varianta A - VS Code** (doporučeno, rozšíření Python a Jupyter už máte z kroku 3):

* otevřete soubor `.ipynb`
* vpravo nahoře vyberte *kernel* -> *Python Environments* -> `.venv` (prostředí
  z kroku 4 - VS Code ho najde automaticky ve vašem adresáři)

**Varianta B - JupyterLab v prohlížeči:**

```
uv run --with jupyterlab jupyter lab
```

* příkaz vypíše adresu (např. `http://localhost:8888/...`), kterou otevřete v prohlížeči
* běžící JupyterLab ukončíte v terminálu klávesami `Ctrl` + `C`

## 10. Slovníček pojmů

**Nástroje a prostředí**

* **Python** - programovací jazyk, který se v kurzu učíme
* **uv** - nástroj pro instalaci Pythonu, balíčků a virtuálních prostředí
* **balíček (package)** - doplňková knihovna kódu, kterou lze k Pythonu přidat (např. NumPy)
* **virtuální prostředí** - oddělené prostředí s vlastními balíčky pro jeden projekt (složka `.venv`)
* **interpret** - program, který čte a vykonává kód Pythonu řádek po řádku
* **terminál / příkazová řádka** - okno, kam se píšou příkazy pro operační systém (ne pro Python)
* **PowerShell** - terminál ve Windows; na macOS a Linuxu se používá Bash nebo Zsh
* **PATH** - seznam cest, kde operační systém hledá spustitelné programy; po instalaci
  je proto potřeba terminál restartovat
* **skript** - textový soubor s příponou `.py` obsahující kód Pythonu
* **editor (VS Code)** - program na psaní a spouštění kódu
* **rozšíření (extension)** - doplněk pro VS Code, který přidává další funkce (např. podporu Pythonu)
* **Command Palette** - vyhledávání příkazů ve VS Code, `Ctrl+Shift+P`

**Notebook**

* **notebook** - soubor `.ipynb`, kde se střídají textové a kódové buňky
* **buňka (cell)** - jedna část notebooku - buď text (markdown), nebo kód
* **markdown** - jednoduchý způsob formátování textu (nadpisy `#`, seznamy `*`, `**tučně**`)
* **kernel** - proces, který v notebooku spouští kód Pythonu
* **magický příkaz (magic command)** - příkaz notebooku začínající `%`, např. `%whos`
* **JupyterLab** - prostředí pro práci s notebooky v prohlížeči
* **Google Colab** - notebook běžící na serverech Google bez nutnosti instalace
* **sezení (session)** - jeden běh prostředí; po jeho ukončení se proměnné
  i doinstalované balíčky ztratí

**Pojmy z Pythonu**

* **proměnná** - jméno, pod kterým je uložena hodnota (např. `delka = 12.5`)
* **přiřazení** - uložení hodnoty do proměnné znakem `=`
* **výraz** - kus kódu, který má hodnotu (např. `2 + 3`)
* **operátor** - znak pro operaci s hodnotami, např. `+`, `-`, `*`, `/`
* **komentář** - text za znakem `#`, který Python ignoruje; slouží čtenáři kódu
* **funkce** - pojmenovaný kus kódu, který se volá jménem a závorkami, např. `print()`
* **datový typ** - druh hodnoty; určuje, co se s hodnotou dá dělat
* **celé číslo (`int`)** - číslo bez desetinné části, např. `25`
* **desetinné číslo (`float`)** - číslo s desetinnou **tečkou**, např. `12.5` (ne `12,5`)
* **řetězec (`str`, string)** - text v uvozovkách, např. `"Dobrý den"`
* **konverze typu** - převod hodnoty na jiný typ, např. `int()`, `float()`
* **`input()`** - funkce pro čtení vstupu z klávesnice; vrací vždy řetězec
* **`print()`** - funkce pro výpis textu na obrazovku
