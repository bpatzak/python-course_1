## 1. Instalace uv

* `uv` je jeden nástroj pro správu Pythonu, balíčků i virtuálních prostředí
* je rychlý a nemusíte nic hledat a instalovat ručně
* dokumentace: <https://docs.astral.sh/uv/>

Instalace na **Windows** (PowerShell):

```
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

Instalace na **macOS / Linux** (terminál):

```
curl -LsSf https://astral.sh/uv/install.sh | sh
```

Po instalaci:

* zavřete a znovu otevřete terminál (aby se načetla nová cesta)
* ověřte, že `uv` funguje:

```
uv --version
```


## 2. Instalace Pythonu pomocí uv

* samotný Python si `uv` stáhne sám, nic nemusíte hledat na webu
* nainstalujte verzi 3.13:

```
uv python install 3.13
```

* seznam dostupných a nainstalovaných verzí:

```
uv python list
```


## 3. Spuštění skriptu `.py`

* vytvořte si adresář pro cvičení, např. `cviceni01`
* v editoru (VS Code) v něm vytvořte soubor `zdravim.py` s obsahem:

```python
jmeno = input("Jak se jmenujete? ")
print("Dobrý den,", jmeno)
```

* v terminálu přejděte do adresáře a skript spusťte:

```
cd cviceni01
uv run zdravim.py
```

Když skript potřebuje nějaký balíček, stačí ho přidat na příkazovou řádku -
`uv` si ho sám stáhne:

```
uv run --with numpy skript.py
```


## 4. Spuštění notebooku `.ipynb`

* notebook je soubor, ve kterém se střídají **textové** a **kódové** buňky
* buňku spustíte klávesami `Shift` + `Enter`
* buňky se spouští v tom pořadí, ve kterém je spustíte vy (ne odshora dolů) -
  na to pozor, je to častý zdroj chyb

**Varianta A - VS Code** (doporučeno):

* nainstalujte rozšíření *Python* a *Jupyter*
* otevřete soubor `.ipynb`
* vpravo nahoře vyberte *kernel* -> *Python Environments* -> vaši verzi Pythonu

**Varianta B - JupyterLab v prohlížeči:**

```
uv run --with jupyterlab jupyter lab
```

* příkaz vypíše adresu (např. `http://localhost:8888/...`), kterou otevřete v prohlížeči
* běžící JupyterLab ukončíte v terminálu klávesami `Ctrl` + `C`
