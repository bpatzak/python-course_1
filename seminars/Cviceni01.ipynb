## 1. Instalace uv

* `uv` je jeden nástroj pro správu Pythonu, balíčků i virtuálních prostředí
* je rychlý a nemusíte nic hledat a instalovat ručně
* dokumentace: <https://docs.astral.sh/uv/>

Instalace na **Windows** (PowerShell):

```
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

Instalace na **macOS / Linux** (terminál):

```
curl -LsSf https://astral.sh/uv/install.sh | sh
```

Po instalaci:

* zavřete a znovu otevřete terminál (aby se načetla nová cesta)
* ověřte, že `uv` funguje:

```
uv --version
```


## 2. Instalace Pythonu pomocí uv

* samotný Python si `uv` stáhne sám, nic nemusíte hledat na webu
* nainstalujte verzi 3.13:

```
uv python install 3.13
```

* seznam dostupných a nainstalovaných verzí:

```
uv python list
```


## 3. Instalace VS Code

VS Code je editor, ve kterém budeme psát skripty i otevírat notebooky.

**Windows / macOS:**

* stáhněte instalátor z <https://code.visualstudio.com/>
* spusťte ho a postupujte podle průvodce (další, další, hotovo)

**Linux (Ubuntu/Debian):**

```
sudo snap install code --classic
```

Po instalaci ještě přidejte dvě rozšíření (záložka *Extensions* vlevo, ikona
čtverečků):

* **Python**
* **Jupyter**

Bez nich VS Code neumí spustit skript ani otevřít notebook.


## 4. Spuštění skriptu `.py`

* vytvořte si adresář pro cvičení, např. `cviceni01`
* v terminálu v něm vytvořte virtuální prostředí (víc v kroku 5):

```
cd cviceni01
uv venv
```

* adresář otevřete v editoru VS Code

```
code .
```

* v editoru v adresáři vytvořte soubor `zdravim.py` s obsahem:

```python

```

* skript spusťte (pokud jste terminál zavřeli, vraťte se do adresáře `cd cviceni01`):

```

```



## 5. Virtuální prostředí

* virtuální prostředí je samostatná složka s vlastní kopií Pythonu a balíčky,
  oddělená od zbytku počítače - typicky jedna na projekt
* už jsme ho v adresáři `cviceni01` vytvořili příkazem `uv venv` (krok 4) -
  vznikla složka `.venv`
* do prostředí přidáme balíček `ipykernel`, který VS Code potřebuje ke
  spouštění notebooků:

```
uv pip install ipykernel
```


## 6. Spuštění notebooku `.ipynb`

* notebook je soubor, ve kterém se střídají **textové** a **kódové** buňky
* buňku spustíte klávesami `Shift` + `Enter`
* buňky se spouští v tom pořadí, ve kterém je spustíte vy (ne odshora dolů) -
  na to pozor, je to častý zdroj chyb

**Varianta A - VS Code** (doporučeno, rozšíření Python a Jupyter už máte z kroku 3):

* otevřete soubor `.ipynb`
* vpravo nahoře vyberte *kernel* -> *Python Environments* -> `.venv` (prostředí
  z kroku 5 - VS Code ho najde automaticky ve vašem adresáři)

**Varianta B - JupyterLab v prohlížeči:**

```
uv run --with jupyterlab jupyter lab
```

* příkaz vypíše adresu (např. `http://localhost:8888/...`), kterou otevřete v prohlížeči
* běžící JupyterLab ukončíte v terminálu klávesami `Ctrl` + `C`

## 10. Slovníček pojmů

**Nástroje a prostředí**

* **Python** - programovací jazyk, který se v kurzu učíme
* **uv** - nástroj pro instalaci Pythonu, balíčků a virtuálních prostředí
* **balíček (package)** - doplňková knihovna kódu, kterou lze k Pythonu přidat (např. NumPy)
* **virtuální prostředí** - oddělené prostředí s vlastními balíčky pro jeden projekt
* **interpret** - program, který čte a vykonává kód Pythonu řádek po řádku
* **terminál / příkazová řádka** - okno, kam se píšou příkazy pro operační systém (ne pro Python)
* **skript** - textový soubor s příponou `.py` obsahující kód Pythonu
* **editor (VS Code)** - program na psaní a spouštění kódu
* **rozšíření (extension)** - doplněk pro VS Code, který přidává další funkce (např. podporu Pythonu)

**Notebook**

* **notebook** - soubor `.ipynb`, kde se střídají textové a kódové buňky
* **buňka (cell)** - jedna část notebooku - buď text (markdown), nebo kód
* **kernel** - proces, který v notebooku spouští kód Pythonu
* **JupyterLab** - prostředí pro práci s notebooky v prohlížeči
* **Google Colab** - notebook běžící na serverech Google bez nutnosti instalace

**Pojmy z Pythonu**

* **proměnná** - jméno, pod kterým je uložena hodnota (např. `delka = 12.5`)
* **konverze typu** - převod hodnoty na jiný typ, např. `int()`, `float()`
* **řetězec (string)** - text v uvozovkách, např. `"Dobrý den"`
* **booleovský výraz** - výraz, jehož výsledkem je `True` nebo `False`
* **`input()`** - funkce pro čtení vstupu z klávesnice
* **`print()`** - funkce pro výpis textu na obrazovku
