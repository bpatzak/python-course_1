## Slovníček pojmů

**Chyby**

* **chybové hlášení (traceback)** - výpis po chybě; čte se odspodu, poslední
  řádek říká typ a příčinu
* **`SyntaxError`** - Python zápisu nerozumí, kód se vůbec nespustil
* **`IndentationError`** - špatné odsazení bloku
* **`NameError`** - použité jméno neexistuje, typicky překlep
* **`TypeError`** - operace nemá smysl pro dané typy
* **`ValueError`** - typ je správný, ale konkrétní hodnota se použít nedá
* **`ZeroDivisionError`** - dělení nulou

**Typy a čísla**

* **datový typ** - druh hodnoty; zjistí ho funkce `type()`
* **celé číslo (`int`)** - číslo bez desetinné části, např. `25`
* **desetinné číslo (`float`)** - číslo s desetinnou **tečkou**, např. `12.5`
* **logická hodnota (`bool`)** - `True` nebo `False`
* **`None`** - hodnota znamenající "nic", chybějící údaj
* **operátor** - znak pro operaci: `+`, `-`, `*`, `/`, `//` (celočíselné dělení),
  `%` (zbytek po dělení), `**` (mocnina)
* **konverze typu** - převod hodnoty na jiný typ: `int()`, `float()`, `str()`
* **`round()`** - zaokrouhlení na daný počet desetinných míst
* **`math.isclose()`** - porovnání desetinných čísel s tolerancí

**Řetězce a výpis**

* **řetězec (`str`)** - text v uvozovkách
* **index** - pozice znaku v řetězci, počítá se od nuly; záporný index od konce
* **slicing** - výřez pomocí `kod[0:5]`
* **neměnnost (immutable)** - řetězec nelze změnit, metody vracejí nový řetězec
* **metoda** - funkce patřící k hodnotě, volá se tečkou: `kod.upper()`
* **f-string** - text s vloženými hodnotami: `f"cena {cena:.2f} Kč"`
* **formátovací zápis** - `:.2f` dvě desetinná místa, `:<12` vlevo, `:>10` vpravo

**Logika a názvosloví**

* **porovnání** - `==`, `!=`, `<`, `>`, `<=`, `>=`; výsledkem je `bool`
* **logický operátor** - `and`, `or`, `not`
* **klíčové slovo** - slovo vyhrazené jazykem, nelze ho použít jako jméno
  (`lambda`, `for`, `if`, ...)
* **snake_case** - zvyk pro jména proměnných: malá písmena a podtržítka
