## Slovníček pojmů

**Soubory**

* **`with open(cesta, rezim, encoding="utf-8") as f:`** - otevře soubor
  a na konci bloku ho zavře
* **režim** - `"r"` čtení, `"w"` zápis s přepsáním, `"a"` připsání na konec
* **kódování** - jak se znaky ukládají do bajtů; `utf-8` je standard,
  `utf-8-sig` pro Excel, `cp1250` starší Windows
* **relativní cesta** - počítá se od složky s notebookem, např. `Data/...`
* **`Path`** - cesta jako objekt; `exists()`, `mkdir()`, `glob()`,
  `read_text()`

**CSV**

* **CSV** - tabulka jako text, jeden řádek na řádek souboru
* **oddělovač** - čárka, v českém Excelu středník: `delimiter=";"`
* **`csv.reader` / `csv.DictReader`** - řádky jako seznamy / slovníky
  řetězců
* **`csv.writer` / `csv.DictWriter`** - zápis řádků ze seznamů / slovníků
* **`newline=""`** - povinné při práci s CSV, jinak vznikají prázdné řádky

**JSON**

* **JSON** - text se slovníky (`{}`) a seznamy (`[]`)
* **`json.load(f)` / `json.dump(data, f)`** - čtení a zápis
* **`indent=2`** - odsazení pro čitelnost
* **`ensure_ascii=False`** - diakritika zůstane čitelná

**Grafy**

* **`fig, ax = plt.subplots()`** - obrázek a osy
* **`ax.plot`, `ax.bar`, `ax.hist`** - čára, sloupce, histogram
* **`label` a `ax.legend()`** - legenda
* **`fig.savefig("graf.png", dpi=200)`** - uložení grafu do souboru
