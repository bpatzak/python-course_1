## Slovníček pojmů

**Kolekce**

* **kolekce** - hodnota, která v sobě drží víc hodnot
* **seznam (`list`)** - uspořádaná měnitelná kolekce, zápis `[1, 2, 3]`
* **`tuple`** - neměnná obdoba seznamu, zápis `(15.2, 8.4)`
* **množina (`set`)** - kolekce jedinečných hodnot bez pořadí, zápis `{1, 2, 3}`
* **slovník (`dict`)** - kolekce dvojic klíč-hodnota, zápis `{"kod": "SO-01"}`
* **klíč** - jméno, pod kterým je hodnota ve slovníku uložená; musí být
  neměnitelný a jedinečný
* **položka (item)** - jedna dvojice klíč-hodnota ve slovníku
* **prvek** - jedna hodnota v seznamu nebo množině
* **vnořená struktura** - kolekce uvnitř kolekce, např. slovník se slovníkem

**Práce s kolekcemi**

* **index** - pozice prvku, počítá se od nuly; záporný index od konce
* **slicing** - výřez, např. `vymery[1:3]` nebo `sorted(vymery)[:3]`
* **agregační funkce** - `len()`, `sum()`, `min()`, `max()`, `sorted()`
* **`sorted()` proti `.sort()`** - `sorted()` vrátí nový seznam, `.sort()` setřídí
  původní a nevrátí nic
* **rozbalení (unpacking)** - `x, y = bod` rozdělí tuple do proměnných
* **`in`** - test, jestli hodnota v kolekci je
* **`.get()`** - přístup ke klíči slovníku s výchozí hodnotou místo chyby
* **`dict(zip(a, b))`** - složení slovníku ze dvou seznamů
* **množinové operace** - `&` průnik, `-` rozdíl, `|` sjednocení

**Objekty a mutabilita**

* **objekt** - hodnota v paměti; proměnná je jen jméno na ni navázané
* **přiřazení nekopíruje** - `b = a` vytvoří druhé jméno pro tentýž objekt
* **`id()`** - identita objektu, **`is`** - jde o tentýž objekt, **`==`** - mají
  stejnou hodnotu
* **měnitelný (mutable)** - objekt lze změnit zevnitř: `list`, `dict`, `set`
* **neměnitelný (immutable)** - změna vyrobí nový objekt: `int`, `float`, `bool`,
  `str`, `tuple`
* **`.copy()`** - plochá kopie, zkopíruje jen vnější kolekci
* **`copy.deepcopy()`** - hluboká kopie, zkopíruje i vnořené kolekce

**Záznam s pevnou strukturou**

* **`dataclass`** - způsob, jak zapsat záznam s pojmenovanými atributy
* **atribut** - pojmenovaná hodnota záznamu, čte se tečkou: `deska.material`
* **anotace typu** - `kod: str`, popis typu pro čtenáře a nástroje; Python ho
  nevynucuje
* **dekorátor** - `@dataclass` nad definicí, doplní hotové chování
