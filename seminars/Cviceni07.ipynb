## Slovníček pojmů

**Argumenty**

* **`*args`** - parametr s hvězdičkou; posbírá libovolný počet pozičních
  argumentů do tuple
* **`**kwargs`** - parametr se dvěma hvězdičkami; posbírá pojmenované argumenty
  do slovníku
* **rozbalení** - hvězdička při volání, `f(*seznam)` nebo `f(**slovnik)`;
  rozloží kolekci na jednotlivé argumenty
* **předání dál** - funkce pošle `**kwargs` beze změny jiné funkci

**Moduly a balíčky**

* **modul** - soubor `.py`; jeho jméno je jméno souboru bez přípony
* **jmenný prostor modulu** - jména uvnitř modulu, dostupná přes tečku,
  `math.pi`
* **`import`** - najde modul, jednou ho spustí a uloží pod jméno
* **`from ... import ...`** - vezme z modulu jen vybraná jména
* **`as`** - jiné, obvykle kratší jméno pro modul, `import statistics as st`
* **docstring modulu** - text v trojitých uvozovkách na prvním řádku souboru
* **balíček** - složka s moduly a souborem `__init__.py`
* **standardní knihovna** - moduly dodávané s Pythonem, nic se neinstaluje
* **zastínění** - váš soubor se jmenuje stejně jako modul ze standardní
  knihovny a `import` načte ten váš
* **třída** - typ, podle kterého se vytvářejí objekty, např. `date` nebo
  `Counter`; zavolání třídy vytvoří objekt

**Skript a notebook**

* **skript** - soubor `.py` spuštěný přímo, `uv run tepelna.py`
* **`__name__`** - jméno modulu; při spuštění jako skript je to `"__main__"`
* **`if __name__ == "__main__":`** - blok, který se provede jen při spuštění
  jako skript, ne při importu
* **kernel** - proces, ve kterém běží kód notebooku; pamatuje si proměnné
  i načtené moduly
* **`importlib.reload()`** - znovu načte změněný modul
* **`%autoreload 2`** - notebook sám načítá změněné moduly před každou buňkou

**Projekt s uv**

* **projekt** - složka s `pyproject.toml`, kódem a prostředím
* **`pyproject.toml`** - popis projektu a seznam balíčků, na kterých závisí
* **`uv.lock`** - přesné verze všech balíčků; podle něj se staví `.venv`
* **`.venv`** - prostředí s nainstalovanými balíčky; dá se kdykoli smazat
  a sestavit znovu
* **`uv add`** - přidá balíček do projektu
* **`uv sync`** - sestaví `.venv` podle `uv.lock`
* **`uv run`** - spustí skript v prostředí projektu
