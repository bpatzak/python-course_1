# Pokusy s želvou
Existuje několik balíčků, které umožňují ovládat želvu z interaktivním notebooku:
- `jupyturtle`, kreslí do notebooku
- `turtle` - kreslí do TK okna


```python
# Balicek nainstalujeme:
# uv pip install jupyturtle
# nebo pokud mame zalozeny uv projekt:
# uv add jupyturtle

import jupyturtle as jt
jt.make_turtle(delay=0.05)
for _ in range(100):
    jt.forward(5)
    jt.left(10)
    jt.toggle_pen()
```

```python
# Balicek nainstalujeme:
# uv pip install turtle
# nebo pokud mame zalozeny uv projekt:
# uv add turtle

# Kresli se do samostatneho TK okna

import turtle

kroky = 1
zmena = 2
uhel = 91
pocet = 20

t = turtle.Turtle()
for _ in range(pocet):
    t.forward(kroky)
    t.right(uhel)
    kroky = kroky + zmena

turtle.done()   # keeps the window responsive until you close it
```
